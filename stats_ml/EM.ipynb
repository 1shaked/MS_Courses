```python
import numpy as np
import random 
```

```python
mus = [0.3 , 0.7]
sigmas = [2 , 1]
ratio = [0.4, 0.6]
```

```python
count = 10
# generate samples based on the prob 
samples = []
for i in range(count):
    p = random.random()
    sum_r = 0
    index = 0
    for i , r in enumerate(ratio):
        sum_r += r
        if sum_r >= p:
            index = i
            break 
    mu = mus[index]
    sigma = sigmas[index]
    value = random.gauss(mu, sigma)
    samples.append(value)
```
